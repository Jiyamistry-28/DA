```python
#count duplicated values
if df.duplicated().sum() != 0:
    df = df.drop_duplicates()
else:
    print("The dataset does not contains duplicate value")
```

```python
#consistent data as data that follows the same rules, formats and logical constriants throughout the dataset. also other posibilites to check consistency?
print(f"Unique season: {df['season'].unique()}")
print(f"Unique month: {df['mnth'].unique()}")
```

```python
#to check the type of dteday column. here the excel automatically convert the dteday column to datetime. but instead, you read it from the csv.
# then the conversion is required.
print(df["dteday"].dtype)
df["dteday"] = pd.to_datetime(df['dteday'])
print(df['dteday'].dtype)
```

```python
#feature selection: it depends on requirements. suppose we want to analyze bike demand.
# we are not requiring instant, casual, or registered columns.
selected_columns = ["dteday","season","yr","mnth","weekday","workingday","weathersit","temp","hum","windspeed","cnt"]
analysis_df = df[selected_columns]
print(analysis_df.columns)
```

```python
def month_map(value):
    dict = {1:"JAN",2:"FEB",3:"MAR",4:"APR",5:"MAY",6:"JUN",7:"JUL",8:"AUG",9:"SEP",10:"OCT",11:"NOV",12:"DEC"}
    return dict[value]
```

```python
analysis_df["year"] = analysis_df["yr"] .apply(lambda year: 2011 if year == 0 else 2012)
analysis_df["month"] = analysis_df["mnth"].apply(month_map)
print(analysis_df["month"].unique())
print(analysis_df["year"].unique())

#Aggregation
total_demand = analysis_df["cnt"].sum() #total demand
print(f"Total demand \n {total_demand}")

year_sum_demand = analysis_df.groupby("year")["cnt"].sum() # year wise total demand
print(f"Year wise demand \n {year_sum_demand}")

yr_demand = analysis_df.groupby("year")["cnt"].mean() # suppose we want avg demand year wise
print(f"Year wise average demand \n {yr_demand}")

yr_mnth_demand = analysis_df.groupby(["year","month"]).mean() # suppose we want avg bike demand by year and month
print(f"Monthly average bike demand by year and month wise \n {yr_mnth_demand}")

mnth_demand = analysis_df.groupby("month")["cnt"].mean() # suppose we want average bike demand by month
print(f"Monthly average bike demand by month wise \n {mnth_demand}")
```

```python
print(f'Total demand of year 2011:\n {analysis_df.loc[analysis_df["year"] == 2011,["cnt"]].sum()}')
print(f'Total active days of year 2011: \n {analysis_df.loc[analysis_df["year"] == 2011,["cnt"]].count()}')
avg_2011 = analysis_df.loc[analysis_df["year"] == 2011, ["cnt"]].sum() / analysis_df.loc[analysis_df["year"] == 2011, ["cnt"]].count()
print(f'Average demand of year 2011: \n {avg_2011}')
print(f'Total demand of April month \n {analysis_df.loc[analysis_df["month"] == "APR", ["cnt"]].sum()}')
print(f'Total active days of April month \n {analysis_df.loc[analysis_df["month"] == "APR",["cnt"]].count()}')
april_avg = analysis_df.loc[analysis_df["month"] == "APR", ["cnt"]].sum() / analysis_df.loc[analysis_df["month"] == "APR",["cnt"]].count()
print(f'Average demand of April Month \n: {april_avg}')
```

```python
import matplotlib.pyplot as plt

# Filter data for 2011
df_2011 = analysis_df[analysis_df["year"] == 2011].copy()

# Plot line chart
plt.figure(figsize=(12, 6))

plt.plot(
    df_2011["dteday"],
    df_2011["cnt"],
    color="blue",
    linewidth=1.5,
    label="Daily Demand"
)

plt.title("Daily Bike Demand in 2011")
plt.xlabel("Date")
plt.ylabel("Total Demand (cnt)")
plt.xticks(rotation=45)
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

april_df = analysis_df[analysis_df["month"] == "APR"].copy()

plt.figure(figsize=(12, 6))

plt.plot(
    april_df["dteday"],
    april_df["cnt"],
    color="green",
    linewidth=2,
    marker="o",
    markersize=3,
    label="April Demand"
)

plt.title("Daily Bike Demand in April")
plt.xlabel("Date")
plt.ylabel("Total Demand (cnt)")
plt.xticks(rotation=45)
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()

plt.tight_layout()
plt.show()
```
