```python
import matplotlib.pyplot as plt # for drawing the graph
from pandas import Series,DataFrame,isna,notna  #for creating the dataframe
```

```python
students = DataFrame({
    "Name" : ["Jay", "Karan", "Dipa", "Jinal", "Krupesh", "Monu"],
    "Id" : [1,2,3,4,5,6],
    "Gender" : ("Male", "Male", "Female", "Female", "Male", "Male"),
    "DA" : (76,89,76,54,38,89),
    "Project" : (56,78,93,78,76,90),
    "ADBMS" : (78,87,56,88,64,89),
    "Course" : ("MCA", "M.sc.IT", "MCA", "M.sc.IT", "M.sc.IT", "M.sc.IT")
})
```

```python
students["Total"] = students.loc[:,["DA","Project","ADBMS"]].sum(axis=1)
students
```

```python
x = students["Name"]
y = students["Total"]
plt.bar(x,y,label="Student Data",color='darkblue')
plt.ylabel("Total Marks")
plt.xlabel("Student Name")
plt.title("Student Marks")
plt.legend()
```

```python
students["Percentage"] = students["Total"] / 3
students
```

```python
x = students["Name"]
y = students["Percentage"]
plt.bar(x,y,label="Student Data",color='grey')
plt.ylabel("Percenages")
plt.xlabel("Student Name")
plt.title("Student Percentage")
plt.legend()
```

```python
maleStudent = students.loc[students["Gender"] == "Male"]
femaleStudent = students.loc[students["Gender"] == "Female"]
x = maleStudent["Name"]
y = maleStudent["Total"]
x1 = femaleStudent["Name"]
y1 = femaleStudent["Total"]
plt.bar(x,y,label="Male Students", color='grey',width=0.5)
plt.bar(x1,y1,label="Female Students", color='darkblue',width=0.5)
plt.ylabel("Total Marks")
plt.xlabel("Student Name")
plt.title("Student Marks")
plt.legend()
```

```python
import numpy as np
x = np.arange(len(students["Name"]));
width = 0.25
# x = [0, 1, 2, 3, 4, 5] = x - 0.25 = 0 - 0.25
daBar = plt.bar(x-width-0.05,students["DA"],width,label="DA",color="darkblue")
projectBar = plt.bar(x,students["Project"],width,label="Project",color="grey")
adbmsBar = plt.bar(x+width+0.05,students["ADBMS"],width,label="ADBMS",color="skyblue")
plt.bar_label(daBar,padding=3)
plt.bar_label(projectBar,padding=3)
plt.bar_label(adbmsBar,padding=3)
plt.legend(
    title = 'Subjects', #title header inside legend
    loc = 'upper right', #placement position upper right, center
    fontsize = 10, #text font size
    frameon = True, #show border size
    facecolor = '#f8f9fa', #legend background color
    edgecolor = 'gray', #legend border color
    ncols = 3 #no of legned columns
)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("DA vs Project marks")
plt.xticks(x,students["Name"])
plt.ylim(0,120)
```

```python
#Data
months = ["January", "February", "March", "April", "May", "June", "July"]
sales = [65, 72, 68, 85, 78, 92, 88]

#find maximum and minimum sales
maximum = max(sales) #92
minimum = min(sales) #65

#find the position of maximum and minimum sales
index1 = sales.index(maximum) #5
index2 = sales.index(minimum) #0

#corresponding month
max_month = months[index1] #june
min_month = months[index2] #january

#create line graph
plt.plot(months, sales, marker='o', linestyle='-.', markersize=5, label="Sales", color='m', markerfacecolor='w', markeredgecolor="b")

#highlight maximum and minimum point
plt.scatter(max_month, maximum, s=100, label="Maximum", color="green", marker='o', alpha=0.5, edgecolors="k")
plt.scatter(min_month, minimum, s=100, label="Minimum", color="blue", marker='o', alpha=0.5, edgecolors="k")

#add title and labels
plt.title("Employee Monthly Sales Performance")
plt.xlabel("Month")
plt.ylabel("Sales")

#add grid and legend
plt.grid(True)
plt.legend()
#plt.show()

#print maximum and minimum information
print("Maximum Sales:", maximum)
print("Sales Month:", max_month)
print("Minimum Sales:", minimum)
print("Sales Month:", min_month)
```
