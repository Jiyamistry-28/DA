```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
#Imputer to handle missing values
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
#to convert categorical variables into numerical values
from sklearn.preprocessing import OneHotEncoder
#to combine preprocessing and model training
from sklearn.pipeline import Pipeline
#to apply different preprocessing and to numerical and categorical columns
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression

#IMPORT EVALUATION METRICS
#to calculate overalll prediction accuracy
from sklearn.metrics import accuracy_score

#to calculate tp,tn,fp,fn
from sklearn.metrics import confusion_matrix

#
from sklearn.metrics import classification_report

#
from sklearn.metrics import roc_auc_score

#to visualize the confusion metrix
from sklearn.metrics import ConfusionMatrixDisplay

#to visualize the ROC Curve
from sklearn.metrics import RocCurveDisplay
```

```python
features = ["pclass","sex","age","sibsp","parch","fare","embarked"]
X = df[features]
y = df["survived"]

print("\nInput Features : ",X.columns)
print("\nTargeted Column : ",y.name)

print("\nX Shape : ",X.shape)
print("\ny Shape : ",y.shape)
```

```python
numeric_features = ["pclass","age","sibsp","parch","fare"]
categorical_feature = ["sex","embarked"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.20,random_state = 42,stratify = y)

print("\nTraining Data Shape : ",X_train.shape)
print("\nTesting Data Shape : ",X_test.shape)
print("\nTraining Target Shape : ",y_train.shape)
print("\nTesting Target Shape : ",y_test.shape)
```

```python
#Numerical Data PreProcessing
numeric_imputer = SimpleImputer(strategy = "median")

numeric_scaler = StandardScaler()

numeric_pipeline = Pipeline(steps = [("imputer",numeric_imputer),("scaler",numeric_scaler)])
```

```python
#Categorical Data PreProcessing
categorical_imputer = SimpleImputer(strategy="most_frequent")

categorical_encoder = OneHotEncoder(handle_unknown = "ignore")

categorical_pipeline = Pipeline(steps = [("imputer",categorical_imputer),("encoder",categorical_encoder)])
```

```python
preprocessor=  ColumnTransformer(
    transformers = [
        ("numeric",numeric_pipeline,numeric_features),
        ("categorical",categorical_pipeline,categorical_feature)
    ]
)
```

```python
logistic_model = LogisticRegression(
    max_iter = 1000
)
```

```python
model = Pipeline(
    steps = [
        ("preprocessor",preprocessor),
        ("classifier",logistic_model)
    ]
)
```

```python
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
```

```python
y_probability = model.predict_proba(X_test)
y_probability_positive = y_probability[:,1]
```

```python
print("\nActual Values")
print(y_test.iloc[:10].values)

print("\nPredicted Values : ")
print(y_pred[:10])

print("\nProbability of Survival : ")
print(y_probability_positive[:10])
```

```python
results = pd.DataFrame({
    "Actual" : y_test.values,
    "Predicted" : y_pred,
    "Survival_probability" : y_probability_positive
})
```
